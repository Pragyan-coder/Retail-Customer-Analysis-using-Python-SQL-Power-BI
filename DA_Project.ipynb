#### Data Conversion: str to object

```python
df_shopping['Gender'] = df_shopping['Gender'].astype('object')
df_shopping['Item Purchased'] = df_shopping['Item Purchased'].astype('object')
df_shopping['Category'] = df_shopping['Category'].astype('object')
df_shopping['Location'] = df_shopping['Location'].astype('object')
df_shopping['Size'] = df_shopping['Size'].astype('object')
df_shopping['Color'] = df_shopping['Color'].astype('object')
df_shopping['Season'] = df_shopping['Season'].astype('object')
df_shopping['Subscription'] = df_shopping['Subscription'].astype('object')
df_shopping['Shipping Type'] = df_shopping['Shipping Type'].astype('object')
df_shopping['Discount Applied'] = df_shopping['Discount Applied'].astype('object')
df_shopping['Promo Code Used'] = df_shopping['Promo Code Used'].astype('object')
df_shopping['Payment Method'] = df_shopping['Payment Method'].astype('object')
df_shopping['Frequency of Purchases'] = df_shopping['Frequency of Purchases'].astype('object')
```

```python
df_shopping.info()
```

```python
df_shopping.describe(include = 'all')
```

#### Dropped 'Anomaly' Column:

```python
df_shopping = df_shopping.drop(['Unnamed: 18'], axis = 1)
```

```python
df_shopping.isnull().sum()
```

#### Removing Null value:

```python
df_shopping['Rating'] = df_shopping.groupby('Category')['Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df_shopping.isnull().sum()
```

```python
df_shopping.columns
```

#### Snake Casing: for uniformity and consistency

```python
df_shopping.columns = df_shopping.columns.str.lower()
df_shopping.columns = df_shopping.columns.str.replace(' ','_')
```

```python
df_shopping.columns
```

```python
df_shopping = df_shopping.rename(columns = {'amt_(usd)':'amt_usd'})
```

#### Creating Column: Age buckets

```python
# create a new column age_group:
labels = ['Young Adult', 'Adult', 'Middle-aged', 'Senior']
df_shopping['age_group'] = pd.qcut(df_shopping['age'], q = 4, labels = labels)
```

```python
df_shopping[['age', 'age_group']].head(5)
```

#### Create a column for 'Frequency Mapping':

```python
# create new column purchase_frequency_days
freq_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}

df_shopping['purchase_freq_days'] = df_shopping['frequency_of_purchases'].map(freq_mapping)
```

```python
df_shopping[['purchase_freq_days','frequency_of_purchases']].head(4)
```

```python
df_shopping.head(3)
```

```python
df_shopping[['discount_applied','promo_code_used']].head(4)
```

#### Dropping unnecessary column:
- 'discount_applied', 'promo_code_used' have the same results, so will drop promo_code_used column!

```python
(df_shopping['discount_applied'] == df_shopping['promo_code_used']).all()
```

```python
df_shopping = df_shopping.drop('promo_code_used', axis = 1)
```

```python
df_shopping.head(3)
```
